# Intelligent Product Review Sentiment Analysis

This notebook reproduces the analysis using the downloaded **Amazon Reviews 2023 — All_Beauty** public review subset. Weak sentiment labels are derived from ratings: 1–2 Negative, 3 Neutral, and 4–5 Positive. These are proxy labels, not manually verified ground truth.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent

reviews = pd.read_csv(root / "data" / "processed" / "processed_reviews.csv")
results = pd.read_csv(root / "reports" / "model_evaluation_results.csv")
print("Processed rows:", len(reviews))
print(reviews["sentiment"].value_counts())
results
```

## Model Comparison

The following metrics are loaded directly from the executed training pipeline. No values below are manually invented.

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(results["model"], results["f1_macro"], color="#2563eb")
ax.set_ylim(0, 1)
ax.set_ylabel("Macro F1-score")
ax.set_title("Actual Model Comparison")
ax.tick_params(axis="x", rotation=20)
plt.tight_layout()
plt.show()
```
